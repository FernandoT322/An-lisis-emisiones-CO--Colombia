# Conclusiones: análisis de las emisiones de CO₂ en Colombia

Este documento resume las conclusiones del análisis de las emisiones anuales de CO₂ de Colombia (1856-2025, 170 registros) con datos de Our World in Data, a partir de los resultados obtenidos en el notebook `Analisis_Emisiones_CO2_Colombia.ipynb`.

## Conclusiones

- Las emisiones anuales de CO₂ de Colombia presentan una tendencia general creciente durante el periodo analizado (1856-2025, 170 registros). El coeficiente de correlación de Pearson entre el año y las emisiones fue de 0,8753, lo que evidencia una relación positiva fuerte.

- Las estadísticas descriptivas muestran una distribución asimétrica: el promedio (aproximadamente 23,28 millones de toneladas) es mucho mayor que la mediana (4,11 millones), y la desviación estándar es de 31,28 millones. Esto indica que las emisiones fueron bajas durante gran parte de la serie y aumentaron de forma marcada en los años más recientes.

- El modelo de regresión lineal obtuvo una pendiente de 556.336,74, es decir, un incremento promedio estimado de aproximadamente 556.337 toneladas de CO₂ por año. Su ecuación es: Y = 556336,74X − 1056290323,0879.

- El coeficiente de determinación (R² = 0,7661) indica que el año explica cerca del 76,61 % de la variabilidad de las emisiones, pero el 23,39 % restante no es capturado por la recta.

- Los residuos tienen una desviación estándar de 15,13 millones de toneladas y van de −22,3 a 36,6 millones, lo que muestra que el modelo lineal no reproduce con precisión las variaciones históricas de las emisiones.

- Al comparar con el modelo polinómico de grado 2, el R² aumenta a aproximadamente 0,9788 y el RMSE disminuye de cerca de 15,1 millones a cerca de 4,5 millones de toneladas. Esto indica que el comportamiento de las emisiones presenta una curvatura que una recta no alcanza a representar.

### Limitaciones

- El modelo utiliza solo el año como variable explicativa, por lo que no considera factores como el crecimiento económico, la población o el consumo de energía.
- Los resultados describen la tendencia observada y no deben usarse para proyecciones futuras sin un análisis adicional.
